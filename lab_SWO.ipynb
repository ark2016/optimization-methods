```julia
using Plots, PlotlyJS, Random, LinearAlgebra

OUTPUT_DIR = "swo_results"
mkpath(OUTPUT_DIR)

POP_SIZE   = 60
MAX_GEN    = 100
MUT_RATE   = 0.15
N_BITS     = 16
SEED       = 42
N_WEIGHTS  = 21

#  Тестовые МКО-задачи 

# ZDT1 — выпуклый фронт: f₂ = 1 - √f₁
function zdt1(x1, x2)
    g = 1.0 + 9.0 * x2
    f1 = x1
    f2 = g * (1.0 - sqrt(f1 / g))
    return f1, f2
end

# ZDT2 — вогнутый фронт: f₂ = 1 - f₁²  (SWO не найдёт вогнутую часть!)
function zdt2(x1, x2)
    g = 1.0 + 9.0 * x2
    f1 = x1
    f2 = g * (1.0 - (f1 / g)^2)
    return f1, f2
end

# SCH — f₁ = x², f₂ = (x-2)²
function sch(x1, _)
    return x1^2, (x1 - 2.0)^2
end

PROBLEMS = [
    (zdt1, (0.0, 1.0),  "ZDT1_выпуклый"),
    (zdt2, (0.0, 1.0),  "ZDT2_вогнутый"),
    (sch,  (-5.0, 5.0), "SCH_простой"),
]

#  Бинарное кодирование 

int_to_bits(x, nb) = digits(x, base=2, pad=nb) |> reverse
bits_to_int(bits) = sum(b * 2^(length(bits)-i) for (i,b) in enumerate(bits))

function float_to_bits(x, xmin, xmax, nb)
    norm = clamp((x - xmin) / (xmax - xmin), 0.0, 1.0)
    iv = round(Int, norm * (2^nb - 1))
    return int_to_bits(iv, nb)
end

function bits_to_float(bits, xmin, xmax)
    iv = bits_to_int(bits)
    return xmin + iv / (2^length(bits) - 1) * (xmax - xmin)
end

encode(x1, x2, xmin, xmax, nb) = vcat(float_to_bits(x1, xmin, xmax, nb), float_to_bits(x2, xmin, xmax, nb))

function decode(chrom, xmin, xmax, nb)
    x1 = bits_to_float(chrom[1:nb], xmin, xmax)
    x2 = bits_to_float(chrom[nb+1:2*nb], xmin, xmax)
    return x1, x2
end

#  Генетические операторы 

function crossover(p1, p2)
    pt = rand(1:length(p1)-1)
    return vcat(p1[1:pt], p2[pt+1:end]), vcat(p2[1:pt], p1[pt+1:end])
end

function mutate!(bits, rate)
    if rand() < rate
        i = rand(1:length(bits))
        bits[i] = 1 - bits[i]
    end
    return bits
end

#  ГА для скалярной задачи: min φ(X) = λ₁f₁(X) + λ₂f₂(X) 

function ga_scalar(mko_func, bounds, λ1, λ2, pop_size, max_gen, mut_rate, nb, seed)
    Random.seed!(seed)
    xmin, xmax = bounds
    pop = [(xmin + rand()*(xmax-xmin), xmin + rand()*(xmax-xmin)) for _ in 1:pop_size]

    # фитнесс = линейная свёртка критериев
    fitness(p) = let (f1, f2) = mko_func(p[1], p[2]); λ1*f1 + λ2*f2 end

    for _ in 1:max_gen
        chroms = [encode(p[1], p[2], xmin, xmax, nb) for p in pop]
        idx = randperm(length(pop))
        offspring = Vector{Vector{Int}}()
        for i in 1:2:length(idx)
            if i+1 > length(idx)
                push!(offspring, copy(chroms[idx[i]]))
                continue
            end
            c1, c2 = crossover(chroms[idx[i]], chroms[idx[i+1]])
            push!(offspring, c1); push!(offspring, c2)
        end
        for ch in offspring; mutate!(ch, mut_rate) end
        offspring_dec = [decode(ch, xmin, xmax, nb) for ch in offspring]
        all_ind = vcat(pop, offspring_dec)
        all_vals = [fitness(p) for p in all_ind]
        si = sortperm(all_vals)
        pop = all_ind[si[1:pop_size]]
    end

    best = pop[1]
    f1, f2 = mko_func(best[1], best[2])
    return best, (f1, f2)
end

#  Фильтрация недоминируемых 

# a доминирует b: ∀k fₖ(a) ≤ fₖ(b) и ∃k fₖ(a) < fₖ(b)
dominates(a, b) = a[1] <= b[1] && a[2] <= b[2] && (a[1] < b[1] || a[2] < b[2])

function filter_nondominated(solutions)
    nd = []
    for (i, si) in enumerate(solutions)
        dominated = any(j != i && dominates(solutions[j][2], si[2]) for j in eachindex(solutions))
        !dominated && push!(nd, si)
    end
    return nd
end

#  SWO 
# Шаг 1: сетка весов λ₁ ∈ [0,1], λ₂ = 1 - λ₁
# Шаг 2: для каждого λ решаем min φ(X) = λ₁f₁ + λ₂f₂
# Шаг 3: собираем все X*, фильтруем недоминируемые → аппроксимация фронта Парето

function run_swo(mko_func, bounds, n_weights, name)
    println("\n" * "="^60)
    println("  SWO: $name | сетка: $n_weights весов")
    println("="^60)

    all_solutions = []

    for i in 0:n_weights-1
        λ1 = i / (n_weights - 1)
        λ2 = 1.0 - λ1
        λ1_eff = max(λ1, 1e-6)
        λ2_eff = max(λ2, 1e-6)

        x_best, f_best = ga_scalar(mko_func, bounds, λ1_eff, λ2_eff,
                                    POP_SIZE, MAX_GEN, MUT_RATE, N_BITS, SEED + i)
        push!(all_solutions, (x_best, f_best, (λ1, λ2)))
        println("  λ=($(round(λ1,digits=2)), $(round(λ2,digits=2))) → f=($(round(f_best[1],digits=4)), $(round(f_best[2],digits=4)))")
    end

    nd = filter_nondominated([(s[1], s[2]) for s in all_solutions])
    println("  Всего: $(length(all_solutions)), недоминируемых: $(length(nd))")
    return all_solutions, nd
end

#  Визуализация 

function save_html(fig, filepath)
    open(filepath, "w") do io
        PlotlyJS.savefig(io, fig, format="html")
    end
end

function plot_pareto_front(all_solutions, nd_solutions, name, mko_func, bounds)
    prefix = lowercase(replace(name, " " => "_"))

    f1_all = [s[2][1] for s in all_solutions]
    f2_all = [s[2][2] for s in all_solutions]
    f1_nd  = [s[2][1] for s in nd_solutions]
    f2_nd  = [s[2][2] for s in nd_solutions]
    si = sortperm(f1_nd)
    f1_nd_s, f2_nd_s = f1_nd[si], f2_nd[si]
    λ_vals = [s[3][1] for s in all_solutions]

    # 1) фронт Парето в пространстве критериев
    traces_front = [
        PlotlyJS.scatter(x=f1_all, y=f2_all, mode="markers",
            marker=attr(size=10, color=λ_vals, colorscale="Viridis",
                        colorbar=attr(title="λ₁"), line=attr(width=1, color="white")),
            name="решения SWO",
            text=["λ₁=$(round(λ,digits=2))" for λ in λ_vals]),
        PlotlyJS.scatter(x=f1_nd_s, y=f2_nd_s, mode="lines+markers",
            line=attr(color="red", width=2, dash="dot"),
            marker=attr(size=6, color="red"),
            name="фронт Парето")
    ]
    fig_front = PlotlyJS.plot(traces_front, PlotlyJS.Layout(
        title="SWO — $name: фронт Парето",
        xaxis=attr(title="f₁(X)"), yaxis=attr(title="f₂(X)"),
        width=750, height=600, template="plotly_white"))
    save_html(fig_front, joinpath(OUTPUT_DIR, "$(prefix)_front.html"))
    println("  → $(prefix)_front.html")
    display(fig_front)

    # 2) геометрия: линии φ = λ₁f₁ + λ₂f₂ = const касаются D_F
    traces_geom = GenericTrace[]

    # сэмплируем область достижимости D_F
    xmin, xmax = bounds
    Random.seed!(123)
    df1, df2 = Float64[], Float64[]
    for _ in 1:5000
        x1 = xmin + rand()*(xmax-xmin)
        x2 = xmin + rand()*(xmax-xmin)
        ff1, ff2 = mko_func(x1, x2)
        push!(df1, ff1); push!(df2, ff2)
    end
    push!(traces_geom, PlotlyJS.scatter(x=df1, y=df2, mode="markers",
        marker=attr(size=2, color="lightgray", opacity=0.3), name="D_F"))

    colors = ["#1f77b4","#ff7f0e","#2ca02c","#d62728","#9467bd","#8c564b","#e377c2"]
    key_λ = [0.0, 0.15, 0.3, 0.5, 0.7, 0.85, 1.0]
    f1r = (minimum(df1), maximum(df1)*1.1)
    f2r = (minimum(df2), maximum(df2)*1.1)

    for (ci, λ1) in enumerate(key_λ)
        λ2 = 1.0 - λ1
        idx = argmin([abs(s[3][1] - λ1) for s in all_solutions])
        sol = all_solutions[idx]
        φ = λ1 * sol[2][1] + λ2 * sol[2][2]

        # линия уровня φ = const: f₂ = (φ - λ₁f₁) / λ₂
        if λ2 > 1e-6
            ff1 = range(f1r[1], f1r[2], length=200)
            ff2 = [(φ - λ1*f) / λ2 for f in ff1]
            valid = findall(f2r[1] .<= ff2 .<= f2r[2])
            if !isempty(valid)
                push!(traces_geom, PlotlyJS.scatter(
                    x=collect(ff1[valid]), y=ff2[valid], mode="lines",
                    line=attr(color=colors[mod1(ci,length(colors))], width=1.5, dash="dash"),
                    name="φ при λ₁=$(round(λ1,digits=2))"))
            end
        end

        push!(traces_geom, PlotlyJS.scatter(
            x=[sol[2][1]], y=[sol[2][2]], mode="markers",
            marker=attr(size=12, color=colors[mod1(ci,length(colors))],
                        symbol="diamond", line=attr(width=2, color="black")),
            name="X*(λ₁=$(round(λ1,digits=2)))"))
    end

    push!(traces_geom, PlotlyJS.scatter(x=f1_nd_s, y=f2_nd_s, mode="lines",
        line=attr(color="red", width=3), name="фронт Парето"))

    fig_geom = PlotlyJS.plot(traces_geom, PlotlyJS.Layout(
        title="SWO — $name: линии φ=λ₁f₁+λ₂f₂=const касаются D_F",
        xaxis=attr(title="f₁", range=collect(f1r)),
        yaxis=attr(title="f₂", range=collect(f2r)),
        width=850, height=700, template="plotly_white"))
    save_html(fig_geom, joinpath(OUTPUT_DIR, "$(prefix)_geometry.html"))
    println("  → $(prefix)_geometry.html")
    display(fig_geom)

    # 3) пространство параметров
    x1_all = [s[1][1] for s in all_solutions]
    x2_all = [s[1][2] for s in all_solutions]
    fig_x = PlotlyJS.plot([
        PlotlyJS.scatter(x=x1_all, y=x2_all, mode="markers",
            marker=attr(size=10, color=λ_vals, colorscale="Viridis",
                        colorbar=attr(title="λ₁"), line=attr(width=1, color="white")),
            name="X*", text=["λ₁=$(round(λ,digits=2))" for λ in λ_vals])
    ], PlotlyJS.Layout(
        title="SWO — $name: пространство параметров",
        xaxis=attr(title="x₁"), yaxis=attr(title="x₂"),
        width=700, height=600, template="plotly_white"))
    save_html(fig_x, joinpath(OUTPUT_DIR, "$(prefix)_params.html"))
    println("  → $(prefix)_params.html")
    display(fig_x)

    # 4) статичный PNG
    p_png = Plots.scatter(f1_all, f2_all,
        xlabel="f₁(X)", ylabel="f₂(X)", title="SWO — $name",
        color=:steelblue, ms=6, label="все решения",
        size=(700, 500), dpi=150, legend=:topright)
    Plots.plot!(p_png, f1_nd_s, f2_nd_s,
        color=:red, lw=2, ls=:dot, marker=:circle, ms=4, label="фронт Парето")
    Plots.savefig(p_png, joinpath(OUTPUT_DIR, "$(prefix)_front.png"))
    println("  → $(prefix)_front.png")
    display(p_png)

    return fig_front, fig_geom
end

#  Сравнение выпуклый vs вогнутый — главная иллюстрация недостатка SWO 

function plot_comparison(res_convex, res_concave)
    nd1, nd2 = res_convex[2], res_concave[2]
    f1_1s = sort([s[2][1] for s in nd1])
    f2_1s = [nd1[i][2][2] for i in sortperm([s[2][1] for s in nd1])]
    f1_2s = sort([s[2][1] for s in nd2])
    f2_2s = [nd2[i][2][2] for i in sortperm([s[2][1] for s in nd2])]

    # истинные фронты
    t = range(0, 1, length=200)
    true_zdt1 = 1.0 .- sqrt.(t)
    true_zdt2 = 1.0 .- t.^2

    fig = PlotlyJS.plot([
        PlotlyJS.scatter(x=collect(t), y=collect(true_zdt1), mode="lines",
            line=attr(color="green", width=2), name="ZDT1 истинный"),
        PlotlyJS.scatter(x=f1_1s, y=f2_1s, mode="markers",
            marker=attr(size=8, color="blue"), name="ZDT1 SWO ✓"),
        PlotlyJS.scatter(x=collect(t), y=collect(true_zdt2), mode="lines",
            line=attr(color="orange", width=2), name="ZDT2 истинный"),
        PlotlyJS.scatter(x=f1_2s, y=f2_2s, mode="markers",
            marker=attr(size=8, color="red", symbol="diamond"), name="ZDT2 SWO ✗")
    ], PlotlyJS.Layout(
        title="SWO: выпуклый фронт ✓  vs  вогнутый ✗",
        xaxis=attr(title="f₁"), yaxis=attr(title="f₂"),
        width=850, height=600, template="plotly_white",
        annotations=[attr(x=0.5, y=0.9, xref="paper", yref="paper",
            text="Линейная свёртка не может найти точки на вогнутой части фронта",
            showarrow=false, font=attr(size=13, color="red"))]))

    save_html(fig, joinpath(OUTPUT_DIR, "comparison_convex_vs_concave.html"))
    println("\n  → comparison_convex_vs_concave.html")
    display(fig)

    # PNG
    p = Plots.plot(collect(t), collect(true_zdt1), lw=2, color=:green, label="ZDT1 истинный",
        xlabel="f₁", ylabel="f₂", title="SWO: выпуклый ✓ vs вогнутый ✗",
        size=(800, 550), dpi=150, legend=:topright)
    Plots.scatter!(p, f1_1s, f2_1s, color=:blue, ms=6, label="ZDT1 SWO")
    Plots.plot!(p, collect(t), collect(true_zdt2), lw=2, color=:orange, label="ZDT2 истинный")
    Plots.scatter!(p, f1_2s, f2_2s, color=:red, ms=6, shape=:diamond, label="ZDT2 SWO")
    Plots.savefig(p, joinpath(OUTPUT_DIR, "comparison.png"))
    println("  → comparison.png")
    display(p)
    return fig
end

#  Запуск 

println("="^60)
println("  МЕТОД ВЗВЕШЕННЫХ КРИТЕРИЕВ (SWO)")
println("  φ(X) = λ₁·f₁(X) + λ₂·f₂(X) → min")
println("  ГА: pop=$POP_SIZE, gen=$MAX_GEN | весов: $N_WEIGHTS")
println("="^60)

all_results = []
for (func, bounds, name) in PROBLEMS
    sol, nd = run_swo(func, bounds, N_WEIGHTS, name)
    plot_pareto_front(sol, nd, name, func, bounds)
    push!(all_results, (sol, nd, name))
end

plot_comparison(all_results[1], all_results[2])

println("\n" * "="^60)
println("  Готово! Файлы в $OUTPUT_DIR/")
println("  *.html — интерактивные PlotlyJS")
println("  *.png — статика")
println("="^60)
```
